````python

import os 
import ants 
import SimpleITK as itk 
````

````python

print(ants.__version__)
print(itk.__version__)
````

````python
# default viewer: wehre we will show image: FJI
image_viwer = itk.ImageViewer()
print(f"default image viwer: {image_viwer}")

# set application: 
image_viwer.SetApplication("/home/yasin/ImageJ/Fiji.app/ImageJ-linux64")
````

````python

paths = os.listdir(path="10604RNIO-L")
single_path = paths[0]
print("Single_Path: {}".format(single_path))
final_path = f"10604RNIO-L/{single_path}"
print("Final path: {}".format(final_path))
img = itk.ReadImage(fileName=final_path,outputPixelType=itk.sitkFloat32)
# print(img), lots of information
print(img.GetOrigin())
print(img.GetSpacing())
print(img.GetDirection())
print(img.GetPixelIDTypeAsString())

# get matrix representation of the pictures:
img_array = itk.GetArrayFromImage(img)
print(type(img_array))
#image_viwer.Execute(img)
````

````python

img2 = itk.ReadImage(fileName=f"10604RNIO-L/{paths[2]}")
print(img2.GetOrigin())
print(img2.GetSpacing())
print(img2.GetDirection())
print(img2.GetPixelIDTypeAsString())
````

````python

images = []
for i in paths:
    full_path = f"10604RNIO-L/{i}"
    #print(full_path)
    images.append(itk.ReadImage(fileName=full_path))
````

````python


# Join into a 3D volume
three_d_image = itk.JoinSeries(images)

# Set spacing: (X_spacing, Y_spacing, slice_thickness)
# Usually X/Y from the pixel spacing of the first slice
x_spacing, y_spacing = images[0].GetSpacing()
slice_thickness = 0.047  # 47 microns in mm

three_d_image.SetSpacing((x_spacing, y_spacing, slice_thickness))

# Save as a 3D volume
itk.WriteImage(three_d_image, "oct_3d.nii")

````
